# Finding 3: solving for the GRU that is the controller

**Finding 3 of three.** Finding 2 showed that training does not produce a GRU that holds the win-stay
rule off the data. This notebook takes the direct route. Instead of training and hoping, it solves by
hand for GRU weights that ARE the controller, exactly, then checks what that buys and what it does
not.

It works with two machines: the two-state win-stay/lose-shift controller (det_init) and a three-state
controller (ts), at hidden sizes H = 1, 2, and 3. Every network here is the analytically **solved**
GRU, not one trained from scratch. The only training is in §5, which starts from the solved weights
and nudges them to ask whether the exact solution is a place training would stay.

Two results carry the finding:

- **A single hidden unit reproduces the three-state machine's behavior exactly.** The catch is that
  at H = 1 you cannot read the memory state back out: the three states collapse onto a line, so no
  exact translation between the GRU's state and the controller's exists. Recovering the state needs
  the hidden size to be at least the number of states minus one (§4).
- **The exact solution holds the off-manifold probe flat**, the test the trained network failed in
  Finding 2 (§3.3). So the drift Finding 2 found is not a limit of the architecture. It is about how
  the weights were arrived at.

## The mapping, done right

Finding 1 left one question open: are the FSC's memory state and the GRU's hidden state two encodings
of the same thing? An earlier attempt answered it by fitting a classifier to guess one from the
other, which scored high and looked convincing. A classifier score hides how far off it actually is.
Section 4 uses the exact tool instead: an **affine map**, a matrix times the state plus an offset. It
either exists, matching to about 1e-8, or it provably cannot, and the condition for which is a clean
rank test. That is the honest replacement for a fitted score.

## How to run it

Run top to bottom from the repository root. The solve is quick numpy and scipy. Section 5 trains
small GRUs from the solved point and loads a cache; nothing else trains, and no committed fit is
overwritten.

## 0. Setup

One cell sets up everything: the imports, a GRU written out in plain numpy so every step is
inspectable, the loaders for the two fitted controllers, and the analytical solver that turns a
controller into GRU weights.

```python
# imports, the numpy GRU, the FSC loaders, and the analytical solver
import sys, os, glob, pickle
import numpy as np
import matplotlib.pyplot as plt
import torch, torch.nn as nn
from torch.func import jacrev
from scipy.optimize import least_squares

torch.set_default_dtype(torch.float64)

# pandas pickle compat shim, so the fitted FSC objects unpickle across versions
import pandas.core.arrays.string_ as _pd_str
from pandas.core.arrays._mixins import NDArrayBacked as _NDArrayBacked
def _compat_string_setstate(self, state):
    if isinstance(state, tuple) and len(state) == 2 and isinstance(state[1], np.ndarray):
        state = (state[0], state[1], {})
    return _NDArrayBacked.__setstate__(self, state)
_pd_str.StringArray.__setstate__ = _compat_string_setstate

REPO    = os.path.abspath(".")
FSC_SRC = os.path.abspath("FSC-inference-MAPSO/src")
if FSC_SRC not in sys.path:
    sys.path.insert(0, FSC_SRC)

# one GRU step in numpy, mirroring torch nn.GRU: h' = (1-z)*n + z*h
def _sigmoid(x):
    return np.where(x >= 0, 1.0 / (1.0 + np.exp(-np.abs(x))),
                    np.exp(-np.abs(x)) / (1.0 + np.exp(-np.abs(x))))
def _softmax(z):
    z = np.asarray(z, float); z = z - z.max(); e = np.exp(z); return e / e.sum()

def gru_step(h, x, P):
    H  = P["H"]
    gi = P["W_ih"] @ x + P["b_ih"]
    gh = P["W_hh"] @ h + P["b_hh"]
    r  = _sigmoid(gi[:H]    + gh[:H])
    z  = _sigmoid(gi[H:2*H] + gh[H:2*H])
    n  = np.tanh( gi[2*H:]  + r * gh[2*H:])
    return (1 - z) * n + z * h, {"reset": r, "update": z, "new": n}

def rnn_action_probs(h, P):
    return _softmax(np.atleast_1d(h) @ P["lin_W"].T + P["lin_b"])

def encode_input(a, y, A, Y):
    x = np.zeros(A + Y); x[a] = 1.0; x[A + y] = 1.0; return x

# teacher-forced hidden trajectory; h[t] is the state the readout uses to predict action t
def rnn_states(P, actions, observations, A, Y):
    h = np.zeros(P["H"]); out = [h.copy()]
    for t in range(len(actions)):
        h, _ = gru_step(h, encode_input(actions[t], observations[t], A, Y), P)
        out.append(h.copy())
    return np.array(out)

def load_dataset(path):
    raw = np.load(os.path.join(REPO, path))
    ra, ro = raw["actions"], raw["observations"]
    ActSpace = np.array(sorted(set(ra.flatten().tolist())))
    ObsSpace = np.array(sorted(set(ro.flatten().tolist())))
    a2i = {v: i for i, v in enumerate(ActSpace)}
    o2i = {v: i for i, v in enumerate(ObsSpace)}
    sessions = [{"actions":      np.array([a2i[a] for a in ra[i]], dtype=int),
                 "observations": np.array([o2i[o] for o in ro[i]], dtype=int)}
                for i in range(ra.shape[0])]
    win = int(np.where(ObsSpace == 1)[0][0]) if 1 in ObsSpace.tolist() else len(ObsSpace) - 1
    return dict(ActSpace=ActSpace, ObsSpace=ObsSpace, A=len(ActSpace), Y=len(ObsSpace),
                win_idx=win, lose_idx=1 - win, sessions=sessions)

# read the fitted FSC as a machine: policy pi(a|m), transitions g(m'|y,a,m), start rho, joint TMat
def load_fsc_machine(name):
    path = glob.glob(os.path.join(REPO, f"fitted_models/{name}/fsc/best_fsc_M*.pkl"))[0]
    with open(path, "rb") as f: fsc = pickle.load(f)
    fsc.set_mode("generation")
    return dict(fsc=fsc, M=int(fsc.M),
                pi=np.asarray(fsc.get_action_policy()),
                g=np.asarray(fsc.get_memory_transitions()),
                rho=np.asarray(fsc.get_rho()),
                TMat=np.asarray(fsc.get_TMat()))

# forward belief filter b_t(m) = P(memory = m | history up to t); b[0] = rho
def fsc_beliefs(TMat, rho, M, actions, observations):
    b = rho.copy(); out = [b.copy()]
    for t in range(len(actions)):
        bn = b @ TMat[int(observations[t]), :, :, int(actions[t])]
        s = bn.sum(); b = bn / s if s > 0 else np.ones(M) / M
        out.append(b.copy())
    return np.array(out)

# every (state, action, observation) the machine can be in; pairs the policy never takes are skipped
def machine_constraints(pi, g, rho, p_min=1e-6):
    M, A = pi.shape; Y = g.shape[0]
    trans = [(m, a, y, g[y, a, m]) for m in range(M) for a in range(A)
             if pi[m, a] > p_min for y in range(Y)]
    return trans, int(np.argmax(rho))

def _shapes(H, A, Y, M):
    return [("W_ih", (3*H, A+Y)), ("b_ih", (3*H,)), ("W_hh", (3*H, H)), ("b_hh", (3*H,)),
            ("lin_W", (A, H)), ("lin_b", (A,)), ("E", (H, M))]

def _unpack(theta, H, A, Y, M):
    out, i = {}, 0
    for k, sh in _shapes(H, A, Y, M):
        n = int(np.prod(sh)); out[k] = theta[i:i+n].reshape(sh); i += n
    return out

# the exact-solve residual (all requirements at the embedded states), reused by both solvers below
def _build_residual(pi, g, rho, H, conf=1e-3):
    M, A = pi.shape; Y = g.shape[0]
    trans, m0 = machine_constraints(pi, g, rho); K = len(trans)
    npar = sum(int(np.prod(sh)) for _, sh in _shapes(H, A, Y, M))
    m_idx = torch.tensor([t[0] for t in trans]); Xb = torch.zeros(K, A + Y)
    for i, (m, a, y, _) in enumerate(trans):
        Xb[i, a] = 1.0; Xb[i, A + y] = 1.0
    Gn      = torch.tensor(np.stack([t[3] for t in trans]))
    log_tgt = torch.tensor(np.log((1.0 - conf) * pi + conf / A))
    def residual_t(theta, lam_t):
        P = _unpack(theta, H, A, Y, M); E = P["E"]; Hb = E[:, m_idx].T
        gi = Xb @ P["W_ih"].T + P["b_ih"]; gh = Hb @ P["W_hh"].T + P["b_hh"]
        r  = torch.sigmoid(gi[:, :H]    + gh[:, :H])
        z  = torch.sigmoid(gi[:, H:2*H] + gh[:, H:2*H])
        n  = torch.tanh(   gi[:, 2*H:]  + r * gh[:, 2*H:])
        r_trans = ((1 - z) * n + z * Hb) - Gn @ E.T
        r_read  = torch.log_softmax(E.T @ P["lin_W"].T + P["lin_b"], 1) - log_tgt
        return torch.cat([r_trans.reshape(-1), r_read.reshape(-1), E[:, m0], lam_t * theta])
    jac_fn = jacrev(residual_t)
    def f(th, lv): return residual_t(torch.tensor(th), torch.tensor(lv)).numpy()
    def J(th, lv): return jac_fn(torch.tensor(th), torch.tensor(lv)).numpy()
    return f, J, npar, M, A, Y

def _unpack_sol(th, H, A, Y, M):
    P = {k: v.numpy().copy() for k, v in _unpack(torch.tensor(th), H, A, Y, M).items()}
    E = P.pop("E"); P["H"] = H
    return P, E

# solve for GRU weights and an embedding E so the GRU IS the machine (pi, g, rho); nothing baked in.
# among restarts, keeps the smallest-weight exact solution (the underdetermined solve's default choice).
def solve_gru_for_fsc(pi, g, rho, H, conf=1e-3, lam=3e-2, restarts=5, seed=0, exact_tol=1e-8):
    f, J, npar, M, A, Y = _build_residual(pi, g, rho, H, conf)
    kw = dict(jac=J, method="lm", xtol=1e-15, ftol=1e-15, gtol=1e-15, max_nfev=5000)
    best, rng = None, np.random.default_rng(seed)
    for k in range(restarts):
        th = 0.5 * rng.standard_normal(npar)
        th = least_squares(f, th, args=(lam,),   **kw).x
        th = least_squares(f, th, args=(1e-12,), **kw).x
        res = float(np.abs(f(th, 0.0)).max()); nrm = float(np.linalg.norm(th))
        key = (res > exact_tol, nrm)
        if best is None or key < best[0]:
            best = (key, res, th)
    _, res, th = best
    P, E = _unpack_sol(th, H, A, Y, M)
    return P, E, res

H_LIST = [1, 2, 3]
DSPEC  = [("det_init", "datasets/wsls_actions_observations_det_init.npz"),
          ("ts",       "datasets/ts_actions_observations.npz")]

DATA = {}
for name, file in DSPEC:
    D   = load_dataset(file)
    FSC = load_fsc_machine(name)
    A, Y = D["A"], D["Y"]
    DATA[name] = dict(name=name, D=D, A=A, Y=Y, WIN=D["win_idx"], LOSE=D["lose_idx"],
                      SESS=D["sessions"], FSC=FSC, M=FSC["M"], pi=FSC["pi"], g=FSC["g"],
                      rho=FSC["rho"], TMat=FSC["TMat"],
                      ACT_NAME={i: f"a={D['ActSpace'][i]:+d}" for i in range(A)},
                      OBS_NAME={D["win_idx"]: "win", D["lose_idx"]: "lose"})

print(f"{'dataset':>10}  {'sessions':>8}  {'trials':>6}  {'M':>3}  {'A':>3}  {'Y':>3}")
for name in DATA:
    d = DATA[name]
    print(f"{name:>10}  {len(d['SESS']):>8}  {len(d['SESS'][0]['actions']):>6}  "
          f"{d['M']:>3}  {d['A']:>3}  {d['Y']:>3}")
```

## 2. Solving at H = 1, 2, 3

The solve pins the network at the controller's memory states: it finds GRU weights and an embedding
(which hidden vector stands for which memory state) so that one GRU step from a state lands exactly
where the controller's table says, and the readout plays exactly what the state says.

The solve is underdetermined: many weight sets are exact at those points. The smallest-weight one, the
obvious default, can be dynamically unstable and drift over real data, so it fails to reproduce the
policy (§3.4). The reference kept here is the smallest-weight exact solution that also reproduces the
policy over data and is invertible where the dimension allows. Sections 3 onward double-check it.

### 2.1 The machine, and what solving means

The controller is three arrays, read straight off the fitted FSC:

| array | shape | what it says |
|---|---|---|
| $\pi(a \mid m)$ | $(M, A)$ | which action state $m$ plays |
| $g(m' \mid y, a, m)$ | $(Y, A, M, M)$ | which state to move to after playing $a$ and seeing outcome $y$ |
| $\rho(m)$ | $(M,)$ | which state it starts in |

A GRU keeps a vector, not a named state, so something has to say which vector means "state $m$".
Call it $\vec{e}_m$ and collect these as the columns of an embedding matrix $E$. The solve finds the
GRU weights **and** $E$ at once, because a placement of the states and a set of weights only work as a
pair.

**One GRU step**, written out (the same equations `gru_step` runs above):

$$\vec{r} = \sigma(W_{ir}\vec{x} + b_{ir} + W_{hr}\vec{h} + b_{hr}), \qquad
\vec{z} = \sigma(W_{iz}\vec{x} + b_{iz} + W_{hz}\vec{h} + b_{hz}),$$

$$\vec{n} = \tanh\!\big(W_{in}\vec{x} + b_{in} + \vec{r}\odot(W_{hn}\vec{h} + b_{hn})\big), \qquad
\vec{h}' = (1-\vec{z})\odot\vec{n} + \vec{z}\odot\vec{h}.$$

Here $\vec{x} = x(a,y)$ is the one-hot encoding of the last action and outcome, $\sigma$ is the
logistic function, and $\odot$ is elementwise. The gate $\vec{z}$ is the blend knob: near 1 it holds
the old state, near 0 it overwrites it with the proposal $\vec{n}$.

**The requirements.** Write one step as $\text{GRU}(\vec{h}, \vec{x})$. The network is the controller
when three conditions hold:

$$\underbrace{\text{GRU}(\vec{e}_m,\, x(a,y)) = \vec{e}_{\,g(y,a,m)}}_{\text{transition, every reachable } (m,a,y)}
\qquad
\underbrace{\operatorname{softmax}(W_{\text{lin}}\vec{e}_m + b_{\text{lin}}) = \pi(\cdot \mid m)}_{\text{readout, every } m}
\qquad
\underbrace{\vec{e}_{m_0} = \vec{0}}_{\text{start}}$$

Land where the table says, play what the state says, and start where tinyRNN starts (its hidden state
is pinned at $\vec{0}$). The transition block carries the weight: it asks the table to hold as an exact
identity in hidden space, on every reachable state, not on average and not only on the trials the data
happened to contain. That exactness is the whole difference from a trained network, which resembles
the table only where the data pinned it.

**One softening.** $\pi$ is a hard one-hot, and a softmax reaches a one-hot only with infinite weights.
So the readout target is pulled a small amount $\texttt{conf} = 10^{-3}$ off the vertex:

$$\tilde{\pi}(\cdot \mid m) = (1 - \texttt{conf})\,\pi(\cdot \mid m) + \tfrac{\texttt{conf}}{A}.$$

This is the only approximation in the construction. $\texttt{conf}$ is how certain the network is
allowed to be, and it sets the $0.9995$ ceiling that shows up later.

**How it is solved.** Each requirement becomes a residual, a number that reads $0$ when the
requirement holds:

$$\vec{r}_{\text{trans}} = \text{GRU}(\vec{e}_m, x(a,y)) - \vec{e}_{g(y,a,m)}, \quad
\vec{r}_{\text{read}} = \log\operatorname{softmax}(W_{\text{lin}}\vec{e}_m + b_{\text{lin}}) - \log\tilde{\pi}(\cdot\mid m), \quad
\vec{r}_{\text{start}} = \vec{e}_{m_0}.$$

Stack them into one long vector and hand it to a least-squares solver (`scipy.optimize.least_squares`
with the exact Jacobian from `torch.func.jacrev`), which searches for the weights and $E$ that drive
every residual to zero together. Two facts shape that search:

- It is **iterative**. Sigmoids and a $\tanh$ sit inside each step, so there is no closed formula; the
  solver steps toward an answer.
- There are **more unknowns than requirements**, so many networks satisfy the list. The solver runs
  from several random starts and keeps the smallest-weight exact solution, which makes the result
  reproducible rather than arbitrary. A tiny weight term $\lambda\,\theta$ is added to the residual to
  express that preference, then relaxed toward zero so the kept solution is still exact.

The code below runs this at $H = 1, 2, 3$ and reports the worst requirement violation at each size.

```python
# mean distance of the hidden trajectory from the FSC belief manifold over data; small = reproduces policy
def track_err_quick(d, P, E, n=40):
    ds = []
    for s in d["SESS"][:n]:
        b = fsc_beliefs(d["TMat"], d["rho"], d["M"], s["actions"], s["observations"])
        h = rnn_states(P, s["actions"], s["observations"], d["A"], d["Y"])
        ds.append(np.linalg.norm(h - b @ E.T, axis=1))
    return float(np.concatenate(ds).mean())
def emb_rank(E, M): return int(np.linalg.matrix_rank(np.hstack([E.T, np.ones((M, 1))]), tol=1e-9))
def replay_res(d, P, E):
    trans, _ = machine_constraints(d["pi"], d["g"], d["rho"])
    return max(float(np.abs(gru_step(E[:, m], encode_input(a, y, d["A"], d["Y"]), P)[0] - E @ gn).max())
               for (m, a, y, gn) in trans)

# keep the smallest-weight exact solution that also reproduces the policy (track < tau); if the default
# already does, keep it; otherwise search varied-scale restarts and prefer invertible, then smallest weight.
def solve_stable(d, H, tau=1e-4):
    pi, g, rho, M, A, Y = d["pi"], d["g"], d["rho"], d["M"], d["A"], d["Y"]
    P0, E0, _ = solve_gru_for_fsc(pi, g, rho, H)
    if track_err_quick(d, P0, E0) < tau:
        return P0, E0
    f, J, npar, _, _, _ = _build_residual(pi, g, rho, H)
    kw = dict(jac=J, method="lm", xtol=1e-15, ftol=1e-15, gtol=1e-15, max_nfev=5000)
    found = []
    for sc in [0.5, 1, 2, 5, 15, 40, 100]:
        for sd in range(5):
            th = sc * np.random.default_rng(int(7919 * sc * 10) + sd).standard_normal(npar)
            th = least_squares(f, th, args=(3e-2,), **kw).x
            th = least_squares(f, th, args=(1e-12,), **kw).x
            if float(np.abs(f(th, 0.0)).max()) < 1e-8:
                P, E = _unpack_sol(th, H, A, Y, M)
                found.append((track_err_quick(d, P, E),
                              max(np.abs(v).max() for k, v in P.items() if k != "H"),
                              emb_rank(E, M), P, E))
    if not found:
        return P0, E0
    stable = [s for s in found if s[0] < tau]
    pool = stable if stable else found
    mr = max(s[2] for s in pool)
    _, _, _, P, E = min([s for s in pool if s[2] == mr], key=lambda s: s[1])
    return P, E

STABLE_CACHE = os.path.join(REPO, "fitted_models", "_stable_solve.pkl")
SOL = {name: {} for name in DATA}
if os.path.exists(STABLE_CACHE):
    _sc = pickle.load(open(STABLE_CACHE, "rb"))
else:
    _sc = {(name, H): solve_stable(DATA[name], H) for name in DATA for H in H_LIST}
    pickle.dump(_sc, open(STABLE_CACHE, "wb"))
for name in DATA:
    for H in H_LIST:
        P, E = _sc[(name, H)]
        SOL[name][H] = (P, E, replay_res(DATA[name], P, E))

print(f"{'dataset':>10}  {'H':>2}  {'max violation':>14}  {'max |weight|':>12}  {'track err':>10}  "
      f"{'rank':>4}  {'invertible':>10}")
for name in DATA:
    d = DATA[name]
    for H in H_LIST:
        P, E, res = SOL[name][H]
        mw = max(np.abs(v).max() for k, v in P.items() if k != "H")
        rk = emb_rank(E, d["M"])
        print(f"{name:>10}  {H:>2}  {res:>14.2e}  {mw:>12.1f}  {track_err_quick(d,P,E):>10.2e}  "
              f"{rk:>4}  {str(rk == d['M']):>10}")
```

## 3. Verification over data

Two questions. First (§3.1): does the solved network produce the controller's policy, meaning do its
action probabilities match the data the controller generated? Second (§3.2): how exactly does it
implement the machine, step by step.

### 3.1 Does the solved network produce the policy?

Teacher-forced: feed the real (action, outcome) histories, read the model's action probabilities, and
compare to the controller's on the same trials.

```python
# teacher-forced: feed the real (action, outcome) histories, read the model's action probabilities,
# and aggregate them into P(repeat action | previous action, previous outcome). Compare data, FSC, solved.
def transition_matrix_data(d, sessions):
    A, Y = d["A"], d["Y"]; num = np.zeros((A, Y, A)); den = np.zeros((A, Y))
    for s in sessions:
        a, o = s["actions"], s["observations"]
        for t in range(1, len(a)):
            num[a[t-1], o[t-1], a[t]] += 1; den[a[t-1], o[t-1]] += 1
    return num / np.maximum(den, 1)[:, :, None]

def transition_matrix_model(d, step_probs, sessions):
    A, Y = d["A"], d["Y"]; num = np.zeros((A, Y, A)); den = np.zeros((A, Y))
    for s in sessions:
        a, o = s["actions"], s["observations"]; p = step_probs(s)
        for t in range(1, len(a)):
            num[a[t-1], o[t-1]] += p[t]; den[a[t-1], o[t-1]] += 1
    return num / np.maximum(den, 1)[:, :, None]

def rates_from_matrix(d, T):
    A, WIN, LOSE = d["A"], d["WIN"], d["LOSE"]
    return (float(np.mean([T[a, WIN, a] for a in range(A)])),
            float(np.mean([1 - T[a, LOSE, a] for a in range(A)])))

def probs_fsc(d):
    return lambda s: fsc_beliefs(d["TMat"], d["rho"], d["M"],
                                 s["actions"], s["observations"])[:-1] @ d["pi"]
def probs_rnn(d, P):
    return lambda s: np.array([rnn_action_probs(hv, P)
                     for hv in rnn_states(P, s["actions"], s["observations"], d["A"], d["Y"])[:-1]])

EVAL_N = 200
for name in DATA:
    d = DATA[name]; ev = d["SESS"][:EVAL_N]
    Td = transition_matrix_data(d, ev)
    models = [("data", None), ("FSC", probs_fsc(d))]
    models += [(f"solved H={H}", probs_rnn(d, SOL[name][H][0])) for H in H_LIST]
    print(f"==== {name} ====")
    print(f"{'model':>12}  {'P(stay|win)':>11}  {'P(shift|lose)':>13}  {'max dev from data':>17}")
    for tag, sp in models:
        T = Td if sp is None else transition_matrix_model(d, sp, ev)
        sw, sl = rates_from_matrix(d, T)
        dev = "" if sp is None else f"{max(abs(T[a,y,a]-Td[a,y,a]) for a in range(d['A']) for y in range(d['Y'])):.4f}"
        print(f"{tag:>12}  {sw:>11.4f}  {sl:>13.4f}  {dev:>17}")
    print()
```

### 3.2 How exactly it implements the machine

Two numbers. **Replay error** is the one-step transition error at the embedded states: does one GRU
step from state m land on state g(m)? **Track error** is the mean distance between the hidden state
and the embedding of the controller's current belief, over real sessions: does the network's memory
sit where the controller's does, trial by trial?

```python
# replay err = one-step transition error at the embedded states; track err = mean distance of the
# hidden state from the FSC belief embedding E@b over real data; CE = predictive cross-entropy per trial.
def ce_fsc(d, sessions):
    tot = n = 0.0
    for s in sessions:
        b = fsc_beliefs(d["TMat"], d["rho"], d["M"], s["actions"], s["observations"])[:-1] @ d["pi"]
        b = np.clip(b, 1e-12, 1.0)
        tot += -np.log(b[np.arange(len(s["actions"])), s["actions"]]).sum(); n += len(s["actions"])
    return tot / n

def ce_rnn(d, P, sessions):
    tot = n = 0.0
    for s in sessions:
        h = rnn_states(P, s["actions"], s["observations"], d["A"], d["Y"])[:-1]
        pr = np.clip(np.array([rnn_action_probs(hv, P) for hv in h]), 1e-12, 1.0)
        tot += -np.log(pr[np.arange(len(s["actions"])), s["actions"]]).sum(); n += len(s["actions"])
    return tot / n

def replay_error(d, P, E):
    trans, _ = machine_constraints(d["pi"], d["g"], d["rho"])
    return max(np.abs(gru_step(E[:, m], encode_input(a, y, d["A"], d["Y"]), P)[0] - E @ gn).max()
               for (m, a, y, gn) in trans)

def track_error(d, P, E, sessions):
    ds = []
    for s in sessions:
        b = fsc_beliefs(d["TMat"], d["rho"], d["M"], s["actions"], s["observations"])
        h = rnn_states(P, s["actions"], s["observations"], d["A"], d["Y"])
        ds.append(np.linalg.norm(h - b @ E.T, axis=1))
    return float(np.concatenate(ds).mean())

for name in DATA:
    DATA[name]["ce_fsc"] = ce_fsc(DATA[name], DATA[name]["SESS"][:EVAL_N])

print(f"{'dataset':>10}  {'H':>2}  {'replay err':>10}  {'track err':>10}  "
      f"{'CE solved':>10}  {'CE FSC':>9}")
for name in DATA:
    d = DATA[name]; ev = d["SESS"][:EVAL_N]
    for H in H_LIST:
        P, E, res = SOL[name][H]
        print(f"{name:>10}  {H:>2}  {replay_error(d,P,E):>10.2e}  {track_error(d,P,E,ev):>10.2e}  "
              f"{ce_rnn(d,P,ev):>10.4f}  {d['ce_fsc']:>9.4f}")
```

```python
# is track error sensitive to how the FSC states are numbered? relabel (flip states 1 and 2) and recompute
def track_relabel(d, P, E, sessions, perm, apply_to_E):
    Ep = E[:, perm] if apply_to_E else E
    ds = []
    for s in sessions:
        b = fsc_beliefs(d["TMat"], d["rho"], d["M"], s["actions"], s["observations"])[:, perm]
        h = rnn_states(P, s["actions"], s["observations"], d["A"], d["Y"])
        ds.append(np.linalg.norm(h - b @ Ep.T, axis=1))
    return float(np.concatenate(ds).mean())

print(f"{'dataset':>10}  {'H':>2}  {'track (as is)':>13}  {'flip 1<->2 (E & belief)':>24}  {'flip belief only':>18}")
for name, H in [("det_init", 1), ("ts", 2)]:
    d = DATA[name]; P, E, _ = SOL[name][H]; ev = d["SESS"][:EVAL_N]; M = d["M"]
    perm = np.arange(M); perm[0], perm[1] = 1, 0
    print(f"{name:>10}  {H:>2}  {track_error(d,P,E,ev):>13.2e}  "
          f"{track_relabel(d,P,E,ev,perm,True):>24.2e}  {track_relabel(d,P,E,ev,perm,False):>18.2e}")
```

Track error does not depend on how the states are numbered. Relabel the states consistently, flipping
states 1 and 2 in both the embedding and the belief, and the number is identical; it moves only under
an inconsistent relabel, which is a bookkeeping error, not a valid relabel. So it measures one thing:
whether the hidden state sits at the embedding of the controller's current belief. That is the right
check for the solved networks here, which sit exactly at their own embedding. It is not used to judge
the retraining sweep in §5.1, where a network slides its representation to an equivalent embedding, so
its distance from the original one grows even while behavior stays perfect.

### 3.3 Off the data manifold: does the rule still hold?

The same probe as Finding 2, applied to the solved networks. Prime each one where it just won on an
arm, then feed nothing but wins on that arm, past the longest win streak the data ever contained.
Win-stay says the only correct curve is flat: keep playing the winning arm. Holding flat out there,
where no training data lives, is the evidence that the network implements the rule rather than fitting
the outputs. This is the test the trained GRU failed in Finding 2.

```python
# win-streak probe: prime at the win-state, feed straight wins past the longest streak, track P(stay)
KMAX = 80
C_H = {1: "#1baf7a", 2: "#2a78d6", 3: "#a35bd1"}

def win_state(d, a0): return int(np.argmax(d["pi"][:, a0]))
def win_h(d, P, a0, n=80):
    hs = []
    for s in d["SESS"][:n]:
        h = np.zeros(P["H"])
        for t in range(len(s["actions"])):
            h, _ = gru_step(h, encode_input(s["actions"][t], s["observations"][t], d["A"], d["Y"]), P)
            if s["actions"][t] == a0 and s["observations"][t] == d["WIN"]: hs.append(h.copy())
    return np.median(np.stack(hs), 0) if hs else np.zeros(P["H"])
def rnn_probe(d, P, a0, kmax=KMAX):
    h = np.asarray(win_h(d, P, a0), float); x = encode_input(a0, d["WIN"], d["A"], d["Y"])
    out = [rnn_action_probs(h, P)[a0]]
    for _ in range(kmax): h, _ = gru_step(h, x, P); out.append(rnn_action_probs(h, P)[a0])
    return np.array(out)
def fsc_probe(d, a0, kmax=KMAX):
    b = np.zeros(d["M"]); b[win_state(d, a0)] = 1.0; out = [float((b @ d["pi"])[a0])]
    for _ in range(kmax):
        bn = b @ d["TMat"][d["WIN"], :, :, a0]; s = bn.sum(); b = bn / s if s > 0 else b
        out.append(float((b @ d["pi"])[a0]))
    return np.array(out)
def max_streak(d):
    m = 0
    for s in d["SESS"]:
        r = 0
        for o in s["observations"]: r = r + 1 if o == d["WIN"] else 0; m = max(m, r)
    return m

kk = np.arange(KMAX + 1)
fig, axes = plt.subplots(len(DATA), 2, figsize=(11.0, 4.2 * len(DATA)), squeeze=False)
DRIFT = {}
for row, name in enumerate(DATA):
    d = DATA[name]; ms = max_streak(d)
    for a0 in range(d["A"]):
        ax = axes[row, a0]
        ax.axvspan(0, ms, color="#8a8a86", alpha=0.13, lw=0)
        ax.text(ms, 0.03, f" longest streak in data (k={ms})", fontsize=7.5, color="0.4",
                ha="left", va="bottom", transform=ax.get_xaxis_transform())
        fp = fsc_probe(d, a0)
        ax.plot(kk, fp, color="#111111", lw=2.6, label="FSC (the rule)", zorder=5)
        for H in H_LIST:
            rp = rnn_probe(d, SOL[name][H][0], a0)
            DRIFT[(name, a0, H)] = (float(np.ptp(rp)), float(np.abs(rp - fp).max()))
            ax.plot(kk, rp, color=C_H[H], lw=1.6, ls="--", label=f"solved H={H}")
        ax.set_ylim(-0.02, 1.05); ax.set_title(f"{name}: win streak on {d['ACT_NAME'][a0]}", fontsize=10)
        ax.set_xlabel("k consecutive wins on the arm")
        if a0 == 0: ax.set_ylabel("P(stay on this arm)")
        ax.legend(fontsize=7.5, loc="lower left")
fig.suptitle("Feed the solved nets nothing but wins, past the longest streak in the data: does P(stay) hold?",
             fontsize=11.5)
plt.tight_layout(); plt.show()

print(f"{'dataset':>10}  {'arm':>6}  {'H':>2}  {'drift (max ptp)':>15}  {'max dev from FSC':>16}")
for name in DATA:
    for a0 in range(DATA[name]["A"]):
        for H in H_LIST:
            dr, dv = DRIFT[(name, a0, H)]
            print(f"{name:>10}  {DATA[name]['ACT_NAME'][a0]:>6}  {H:>2}  {dr:>15.2e}  {dv:>16.2e}")
```

Every solved net stays on the controller's line to within about 1e-8 across 80 straight wins, roughly
45 past the longest streak the data contained. A network that had only fit the outputs would have no
reason to hold out there and would peel off the rule as k leaves the shaded region; these do not. The
probe holds flat at every size, so it does not separate H = 1 from the larger ones here: a pure win
streak sits in a stable win-state for every size, and the drift that separates them shows up only on
the full trajectories of §3.2, which pass through losses and the stochastic state.

### 3.4 The selection: why not the smallest-weight solve

```python
# the smallest-weight exact solve (the default) is exact at the constraint points but can be dynamically
# unstable. drift is a trajectory property: perturb the hidden state mid-session and measure how the
# dynamics amplify it along the real path (a single fixed-point eigenvalue does not predict it).
_rng = np.random.default_rng(0)
def perturb_growth(d, P, t0=100, eps=1e-8, n=40):
    A, Y = d["A"], d["Y"]; worst = 0.0
    for s in d["SESS"][:n]:
        acts, obs = s["actions"], s["observations"]; T = len(acts)
        h = np.zeros(P["H"])
        for t in range(t0): h, _ = gru_step(h, encode_input(acts[t], obs[t], A, Y), P)
        v = _rng.standard_normal(P["H"]); hp = h + eps * v / np.linalg.norm(v); hb = h.copy()
        for t in range(t0, T):
            x = encode_input(acts[t], obs[t], A, Y)
            hb, _ = gru_step(hb, x, P); hp, _ = gru_step(hp, x, P)
        worst = max(worst, float(np.linalg.norm(hp - hb) / eps))
    return worst

# the default smallest-weight solve for the two sizes the selection had to repair
DEFAULT = {("ts", H): solve_gru_for_fsc(DATA["ts"]["pi"], DATA["ts"]["g"], DATA["ts"]["rho"], H)[:2]
           for H in [2, 3]}
d = DATA["ts"]; ev = d["SESS"][:60]
print(f"ts:  the smallest-weight default vs the selected reference\n")
print(f"{'H':>2}  {'solution':>26}  {'max|weight|':>11}  {'track err':>10}  {'perturb growth':>15}")
for H in [2, 3]:
    for tag, (P, E) in [("default (smallest weight)", DEFAULT[("ts", H)]),
                        ("selected reference", (SOL["ts"][H][0], SOL["ts"][H][1]))]:
        mw = max(np.abs(v).max() for k, v in P.items() if k != "H")
        print(f"{H:>2}  {tag:>26}  {mw:>11.1f}  {track_error(d,P,E,ev):>10.1e}  {perturb_growth(d,P):>15.1e}")
```

```python
# same policy test as 3.1, ts H=2 and H=3: the smallest-weight default vs the selected reference
d = DATA["ts"]; ev = d["SESS"][:EVAL_N]; Td = transition_matrix_data(d, ev)
print(f"{'H':>2}  {'solution':>26}  {'P(stay|win)':>11}  {'P(shift|lose)':>13}  {'CE/trial':>9}")
sw, sl = rates_from_matrix(d, Td)
print(f"{'':>2}  {'data':>26}  {sw:>11.4f}  {sl:>13.4f}  {'--':>9}")
for H in [2, 3]:
    for tag, P in [("default (smallest weight)", DEFAULT[("ts", H)][0]),
                   ("selected reference", SOL["ts"][H][0])]:
        T = transition_matrix_model(d, probs_rnn(d, P), ev); sw, sl = rates_from_matrix(d, T)
        print(f"{H:>2}  {tag:>26}  {sw:>11.4f}  {sl:>13.4f}  {ce_rnn(d,P,ev):>9.4f}")
```

The smallest-weight default is not a solve failure: at H = 2 and H = 3 it is exact at the constraint
points to 1e-16. What it lacks is dynamical stability. Perturb the hidden state mid-session and follow
the real path: a reference that reproduces the policy damps the perturbation to nothing, while the
smallest-weight default at ts H = 2 and H = 3 amplifies it by thousands, so its state drifts off the
belief manifold and its policy comes out wrong. This is a property of the whole trajectory, not of one
step: det_init H = 2 has an expanding one-step Jacobian, largest eigenvalue about 2, yet still damps
along its real path, so the single-step eigenvalue is the wrong thing to read. The fix is to select on
stability, not on weight: among the exact solutions, keep the smallest-weight one that reproduces the
policy and is invertible. That solution is not extreme. For ts it sits at weight 23 (H = 2) and 5
(H = 3), the latter smaller than the drifting default, so stability is about which exact solution the
search lands on, not how large the weights are.

## 4. The hidden representation, and the map done right

The right way to ask whether the GRU's hidden state encodes the controller's memory is to compare
against a network we have proven is the controller. §3 showed the solved network is exactly that, so
its hidden state is the faithful encoding of the FSC state, and the question becomes whether an exact
map ties the two together. The wrong way, which Finding 1 flagged and set aside, is to fit a
classifier that guesses one state from the other and report how often it is right: a high score can
sit on top of large reconstruction error, so it reports that the information is present without
showing the two states correspond. §4.1 asks the weak question, is the state recoverable at all. §4.2
asks the strict one, is there an affine map that returns the controller's state from the hidden vector
so the two spaces line up exactly. For the solved network the answer is yes to about 1e-8 where the
dimension allows, and provably no below it, with a rank test saying which. A trained network fails
this same map even where the dimension allows it (the experiments notebook measures the error), which
is the precise sense in which it does not hold the controller's representation.

### 4.1 Decode on the readout axis

The controller's state, read from the solved network's hidden state, projected onto the one axis the
action is actually read from: the readout logit difference. Colors that separate mean the memory state
is recoverable from what the network outputs.

```python
# project each solved hidden state onto the readout logit difference, colour by the true FSC state
def logit_axis(d, P, sessions):
    w = P["lin_W"][0] - P["lin_W"][1]; b = P["lin_b"][0] - P["lin_b"][1]; out = []
    for s in sessions:
        for h in rnn_states(P, s["actions"], s["observations"], d["A"], d["Y"])[:-1]:
            out.append(float(w @ np.atleast_1d(h) + b))
    return np.array(out)

def fsc_labels(d, sessions):
    return np.concatenate([fsc_beliefs(d["TMat"], d["rho"], d["M"],
                           s["actions"], s["observations"])[:-1].argmax(1) for s in sessions])

PANELS = [("det_init", 1), ("ts", 1), ("ts", 2)]
fig, axes = plt.subplots(1, len(PANELS), figsize=(4.6 * len(PANELS), 3.6))
for ax, (name, H) in zip(axes, PANELS):
    d = DATA[name]; P, E, _ = SOL[name][H]; ss = d["SESS"][:120]
    v = logit_axis(d, P, ss); lab = fsc_labels(d, ss)
    bins = np.linspace(v.min() - 0.5, v.max() + 0.5, 200)
    for m in range(d["M"]):
        if (lab == m).any():
            ax.hist(v[lab == m], bins=bins, color=C_STATE[m], alpha=0.72, label=f"FSC m={m}")
    ax.set_yscale("log"); ax.set_title(f"solved {name}  H={H}", fontsize=10)
    ax.set_xlabel("readout logit difference"); ax.legend(fontsize=8)
axes[0].set_ylabel("trials (log)")
fig.suptitle("Solved GRU: the FSC memory state read from the hidden state, on the action-readout axis",
             fontsize=11)
plt.tight_layout(); plt.show()

print(f"{'dataset':>10}  {'H':>2}  {'state recovered on axis':>24}")
for name, H in PANELS:
    d = DATA[name]; P, E, _ = SOL[name][H]; ss = d["SESS"][:120]
    v = logit_axis(d, P, ss); lab = fsc_labels(d, ss)
    meds = np.array([np.median(v[lab == m]) if (lab == m).any() else np.inf for m in range(d["M"])])
    acc = (np.argmin(np.abs(v[:, None] - meds[None, :]), axis=1) == lab).mean()
    print(f"{name:>10}  {H:>2}  {acc:>23.4%}")
```

### 4.2 Invertibility of the embedding: the affine map

Decoding on the readout axis asks whether the state is recoverable at all. Invertibility asks the
stricter question: is there an **affine** map, A h + b, that returns the controller's memory state from
the hidden vector, so the two state spaces are in exact correspondence. Solving A e_m + b = one-hot(m)
for the M embedded states needs the matrix [E^T | 1] to have rank M, which requires **H >= M - 1**.
Below M - 1 there is no such map, no matter how the weights are set. This is the affine mapping, and
its condition is the whole story: matching the hidden size to the number of states is not the
requirement it looks like; one less is enough, and one less than that is impossible.

For a trained network the same map, fit on its hidden states, carries real reconstruction error (experiments/analytical_rnn_fsc_wsls.ipynb §6 finds a worst miss of 0.4 to 1.5 sitting behind an R^2 near 0.99), so a good classifier score does not mean the two states correspond. The H >= M - 1 condition is the low-hidden-unit limit itself: at H = 1 the three-state machine's states collapse onto a line and no affine map recovers them; at H = 2 they spread into a plane and the map inverts.

```python
# fit the affine inverse A h + b = one-hot(m) from the embedded states; report its rank and error
def affine_inverse_from_embedding(E):
    Hh, Mm = E.shape
    S = np.hstack([E.T, np.ones((Mm, 1))])
    Theta, *_ = np.linalg.lstsq(S, np.eye(Mm), rcond=None)
    rank = int(np.linalg.matrix_rank(S, tol=1e-9))
    exact = float(np.abs(S @ Theta - np.eye(Mm)).max())
    return rank, exact

print(f"{'dataset':>10}  {'H':>2}  {'M':>2}  {'needs H>=M-1':>12}  {'rank[E^T|1]':>11}  "
      f"{'invertible':>10}  {'recon err':>10}")
for name in DATA:
    d = DATA[name]; M = d["M"]
    for H in H_LIST:
        P, E, _ = SOL[name][H]
        rank, err = affine_inverse_from_embedding(E)
        print(f"{name:>10}  {H:>2}  {M:>2}  {str(H >= M-1):>12}  {rank:>11}  "
              f"{str(rank == M):>10}  {err:>10.2e}")
```

```python
# the embedded states in hidden space: a line for H=1, a plane for H=2, with the min gap between states
fig, axes = plt.subplots(1, len(PANELS), figsize=(4.6 * len(PANELS), 3.6))
for ax, (name, H) in zip(axes, PANELS):
    d = DATA[name]; P, E, _ = SOL[name][H]; M = d["M"]
    rank, _ = affine_inverse_from_embedding(E)
    verdict = "invertible" if rank == M else "NOT invertible"
    if H == 1:
        for m in range(M):
            ax.scatter(E[0, m], 0.0, s=170, color=C_STATE[m], zorder=3)
            ax.annotate(f"m{m}\n{E[0,m]:.1e}", (E[0, m], 0.0), textcoords="offset points",
                        xytext=(0, 14 + 22 * m), ha="center", fontsize=8, color=C_STATE[m])
        gap = min(abs(E[0, i] - E[0, j]) for i in range(M) for j in range(i+1, M))
        pad = max(3 * gap, 1e-3)
        ax.set_xlim(E[0].min() - pad, E[0].max() + pad); ax.set_ylim(-1, 1)
        ax.set_yticks([]); ax.set_xlabel("hidden value h")
    else:
        for m in range(M):
            ax.scatter(E[0, m], E[1, m], s=170, color=C_STATE[m], zorder=3)
            ax.annotate(f"m{m}", (E[0, m], E[1, m]), textcoords="offset points",
                        xytext=(8, 8), fontsize=9, color=C_STATE[m])
        gap = min(np.linalg.norm(E[:, i] - E[:, j]) for i in range(M) for j in range(i+1, M))
        ax.set_xlabel("h0"); ax.set_ylabel("h1")
    ax.set_title(f"solved {name}  H={H}\nmin gap {gap:.1e},  rank {rank}/{M},  {verdict}", fontsize=9.5)
fig.suptitle("Where the solved network places the FSC states in hidden space", fontsize=11)
plt.tight_layout(); plt.show()
```

The three ts states at H = 1 sit within about 1e-3 of each other on a single line, and the rank of
[E^T | 1] is 2, one short of the M = 3 needed, so no affine map recovers the memory state: the
embedding is not invertible. The network still predicts correctly because the readout is a softmax,
which is nonlinear and can turn three points on a line into three different action distributions. So a
one-unit GRU reproduces the three-state machine's behavior while failing to hold its memory as a
recoverable state. At H = 2 the three states spread into the plane, the rank is 3, and the map
inverts. This is the finding: behavior transfers to a single unit, but the correspondence between the
two memories needs at least M - 1 units, and no amount of training changes that.

## 5. Stability from the analytical initialization, and why training misses the solution

Section 4 is about the solved network as it stands. This section starts from the solved weights and
trains, to ask whether the exact solution is a place training stays. The metrics are cross-entropy
over data and the off-manifold rule deviation, because the win-streak probe is saturated at H = 1 and
would read a dead network as a healthy one.

Section 5.1 sweeps the learning rate from the exact point at L1 = 0. Section 5.2 knocks the weights
off and asks whether training walks back. Section 5.3 adds the piece that names the culprit: retrain
from the exact point with the L1 penalty the committed fits used and the rule erodes, retrain with the
penalty off and it holds, and start cold and training never finds the rule at all. Cross-entropy has
no gradient off the data, so the exact solution is one of infinitely many networks that fit the data
equally, and nothing in training selects it. The fuller basin sweep, how close a start has to sit for
training to walk back, is in experiments/fsc_rnn_wsls_paper.ipynb §5.7.

```python
# torch bridge to and from the numpy weight dict, plus the batched data and the training harness
class TinyGRU(nn.Module):
    def __init__(self, Hd, A, Y):
        super().__init__(); self.rnn = nn.GRU(A + Y, Hd); self.lin = nn.Linear(Hd, A); self.Hd = Hd
    def forward(self, X):
        h0 = torch.zeros(1, X.shape[1], self.Hd, dtype=X.dtype)
        out, _ = self.rnn(X, h0); out = torch.cat((h0, out), 0); return self.lin(out)[:-1]

def to_torch(P, A, Y):
    m = TinyGRU(P["H"], A, Y); sd = m.state_dict()
    sd["rnn.weight_ih_l0"] = torch.tensor(P["W_ih"]); sd["rnn.weight_hh_l0"] = torch.tensor(P["W_hh"])
    sd["rnn.bias_ih_l0"] = torch.tensor(P["b_ih"]);  sd["rnn.bias_hh_l0"] = torch.tensor(P["b_hh"])
    sd["lin.weight"] = torch.tensor(P["lin_W"]);     sd["lin.bias"] = torch.tensor(P["lin_b"])
    m.load_state_dict(sd); return m

def from_torch(m):
    p = {k: v.detach().numpy() for k, v in m.rnn.named_parameters()}
    return {"W_ih": p["weight_ih_l0"], "W_hh": p["weight_hh_l0"],
            "b_ih": p["bias_ih_l0"], "b_hh": p["bias_hh_l0"],
            "lin_W": m.lin.weight.detach().numpy(), "lin_b": m.lin.bias.detach().numpy(), "H": m.Hd}

def batch(d, n):
    A, Y = d["A"], d["Y"]; S = d["SESS"][:n]; T = len(S[0]["actions"])
    X = np.zeros((T, len(S), A + Y)); Yt = np.zeros((T, len(S)), dtype=np.int64)
    for j, s in enumerate(S):
        for t in range(T):
            X[t, j, s["actions"][t]] = 1.0; X[t, j, A + s["observations"][t]] = 1.0
            Yt[t, j] = s["actions"][t]
    return torch.tensor(X), torch.tensor(Yt)

# off-manifold rule deviation: how far the net's P(stay) sits from the FSC's, over a win streak past
# the data. behaviour (not geometry), so it neither false-alarms on representation drift nor hides
# behind a frozen hidden state; win_h uses fewer sessions here to keep the sweep fast.
def rule_dev(d, P, kmax=KMAX):
    worst = 0.0
    for a0 in range(d["A"]):
        h = np.asarray(win_h(d, P, a0, n=40), float); x = encode_input(a0, d["WIN"], d["A"], d["Y"])
        rp = [rnn_action_probs(h, P)[a0]]
        for _ in range(kmax): h, _ = gru_step(h, x, P); rp.append(rnn_action_probs(h, P)[a0])
        worst = max(worst, float(np.abs(np.array(rp) - fsc_probe(d, a0, kmax)).max()))
    return worst

# train from P0; at each snapshot record CE over data, track error vs the fixed E, and rule deviation
def train_track(d, P0, E, lr, epochs, every, n_train, eval_sess):
    A, Y = d["A"], d["Y"]; X, Yt = batch(d, n_train); m = to_torch(P0, A, Y)
    opt = torch.optim.AdamW(m.parameters(), lr=lr, weight_decay=0.0); lossf = nn.CrossEntropyLoss()
    ep, ce, tr, rd = [], [], [], []
    for e in range(epochs + 1):
        if e % every == 0:
            Pn = from_torch(m)
            ep.append(e); ce.append(ce_rnn(d, Pn, eval_sess)); tr.append(track_error(d, Pn, E, eval_sess))
            rd.append(rule_dev(d, Pn))
        opt.zero_grad(); loss = lossf(m(X).reshape(-1, A), Yt.reshape(-1)); loss.backward(); opt.step()
    return np.array(ep), np.array(ce), np.array(tr), np.array(rd)

def flat(P): return np.concatenate([np.asarray(v).ravel() for k, v in P.items() if k != "H"])
def perturb(P, eps, seed):
    rng = np.random.default_rng(seed)
    return {k: (v if k == "H" else v + eps * rng.standard_normal(np.asarray(v).shape))
            for k, v in P.items()}
```

The top row is cross-entropy minus the controller's own cross-entropy, so 0 means the network matches

controller's, which is 1 for win-stay. Zero means the network still implements the rule where no data
lives; it rises only when the rule breaks. This reads correctly on both machines where track error did
not. On det_init it stays near zero, because behavior is preserved even as the representation slides.
On ts it jumps to 1 exactly when a high learning rate ejects the solution and the readout stops obeying
win-stay. It works because it is behavior, not geometry, so det_init's representation drift does not
move it, and because it reads the readout, not the frozen hidden state, so ts's ejection cannot hide
from it. It also carries information cross-entropy does not: a network can fit the data yet break the
rule off the data, and this is the test that separates implementing the controller from imitating it.

### 5.2 Perturbation ladder: knock the weights off, then retrain

Run on det_init, the two-state case. Take the solved weights, add eps times random noise, retrain at
L1 = 0, and ask whether the optimizer returns to a network that reproduces the controller. The control
is the same optimizer and budget started from a random init, the way an ordinary fit begins. If the
random control sits well below the near-the-solution runs, the exact rule is reachable only from close
by, and a cold start lands outside its basin. ts at H = 1 is not run here: its solved weights reach
about 5689, a saturated needle in weight space, so any retraining either cannot move it or ejects it
outright, and a fixed-size perturbation is negligible against weights that large.

```python
# det_init: theta0 = theta* + eps*xi, retrain, verdict by CE over data; random-init is the control
PB_NAME = "det_init"
PLR, P_EPOCHS, P_SEEDS = 5e-4, 1200, [0, 1, 2]
EPS_LADDER, N_RANDOM = [0.0, 0.5, 1.0, 2.0, 4.0], 3
HOLD_CE = DATA[PB_NAME]["ce_fsc"] + 0.05

PB_CACHE = os.path.join(REPO, "fitted_models", "_perturb_cache.pkl")
PB_SIG = dict(name=PB_NAME, lr=PLR, epochs=P_EPOCHS, seeds=P_SEEDS, eps=EPS_LADDER, nrand=N_RANDOM,
              n=TRAIN_N, evals=EVAL_S, H=LR_H)
PBRUN = None
if os.path.exists(PB_CACHE):
    with open(PB_CACHE, "rb") as f: blob = pickle.load(f)
    if blob.get("sig") == PB_SIG: PBRUN = blob["runs"]
if PBRUN is None:
    d = DATA[PB_NAME]; Pstar, E, _ = SOL[PB_NAME][LR_H]; th0 = flat(Pstar); nrm = np.linalg.norm(th0)
    PBRUN = []
    for eps in EPS_LADDER:
        for sd in (P_SEEDS if eps > 0 else [0]):
            P0 = Pstar if eps == 0 else perturb(Pstar, eps, sd)
            _, ce, tr, _rd = train_track(d, P0, E, PLR, P_EPOCHS, P_EPOCHS, TRAIN_N, _eval[PB_NAME])
            PBRUN.append(dict(kind="solved", eps=eps, seed=sd,
                              rel=float(np.linalg.norm(flat(P0) - th0) / nrm),
                              ce_end=float(ce[-1]), tr_end=float(tr[-1])))
    for sd in range(N_RANDOM):
        torch.manual_seed(1000 + sd); Pr = from_torch(TinyGRU(LR_H, d["A"], d["Y"]).double())
        _, ce, tr, _rd = train_track(d, Pr, E, PLR, P_EPOCHS, P_EPOCHS, TRAIN_N, _eval[PB_NAME])
        PBRUN.append(dict(kind="random", eps=float("nan"), seed=sd,
                          rel=float(np.linalg.norm(flat(Pr) - th0) / nrm),
                          ce_end=float(ce[-1]), tr_end=float(tr[-1])))
    with open(PB_CACHE, "wb") as f: pickle.dump(dict(sig=PB_SIG, runs=PBRUN), f)
print("done")
```

```python
# left: fraction of runs reproducing the FSC vs eps; right: CE@end vs relative perturbation
d = DATA[PB_NAME]; floor = d["ce_fsc"]
sol = [r for r in PBRUN if r["kind"] == "solved"]; ran = [r for r in PBRUN if r["kind"] == "random"]
fig, axes = plt.subplots(1, 2, figsize=(12.0, 4.2))

ax = axes[0]
fr = [(e, np.mean([r["ce_end"] <= HOLD_CE for r in sol if r["eps"] == e])) for e in EPS_LADDER]
ax.plot([e for e, _ in fr], [f for _, f in fr], "o-", color="#1baf7a", lw=2.0, ms=6,
        label="from the solved weights")
ax.axhline(np.mean([r["ce_end"] <= HOLD_CE for r in ran]), color="#eb6834", ls="--", lw=1.5,
           label="from random init")
ax.set_xlabel(r"perturbation size $\varepsilon$"); ax.set_ylim(-0.05, 1.08)
ax.set_ylabel("fraction of runs reproducing the FSC"); ax.set_title("Does training walk back?", fontsize=10.5)
ax.legend(fontsize=8, loc="best")

ax = axes[1]
ax.scatter([max(r["rel"], 1e-3) for r in sol], [r["ce_end"] - floor for r in sol], s=60,
           color="#1baf7a", zorder=3, label="from the solved weights")
ax.scatter([max(r["rel"], 1e-3) for r in ran], [r["ce_end"] - floor for r in ran], s=60, marker="X",
           color="#eb6834", zorder=3, label="from random init")
ax.axhline(HOLD_CE - floor, color="0.4", ls="--", lw=1.2,
           label=f"reproduces the FSC (below {HOLD_CE - floor:.2f})")
ax.set_xscale("log"); ax.set_yscale("symlog", linthresh=0.05)
ax.set_yticks([0, 0.05, 0.5, 5]); ax.set_yticklabels(["0", "0.05", "0.5", "5"]); ax.set_ylim(-0.02, 8)
ax.set_xlabel(r"relative perturbation $\||\Delta\theta\||/\||\theta^*\||$")
ax.set_ylabel("CE after retraining minus FSC CE   (0 = matches; log above 0.05)"); ax.set_title("Where each run ended up", fontsize=10.5)
ax.legend(fontsize=8, loc="best")
fig.suptitle(f"{PB_NAME}: perturb the solved weights, retrain, and ask whether win-stay comes back",
             fontsize=11.5)
plt.tight_layout(); plt.show()

print(f"{'start':>8}  {'eps':>5}  {'seed':>4}  {'rel dtheta':>10}  {'CE@end':>8}  "
      f"{'track@end':>10}  {'verdict':>8}")
for r in PBRUN:
    eps = "rand" if r["kind"] == "random" else f"{r['eps']:g}"
    ok = "holds" if r["ce_end"] <= HOLD_CE else "broken"
    print(f"{r['kind']:>8}  {eps:>5}  {r['seed']:>4}  {r['rel']:>10.3f}  {r['ce_end']:>8.4f}  "
          f"{r['tr_end']:>10.2e}  {ok:>8}")
```

### 5.3 The L1 penalty erodes the exact solution, and a cold start never finds it

This section and the next gather the whole story of the L1 regularizer in one place. §5.3 works from
the exact analytical solution; §5.4 turns to the committed trained networks.

Section 5.1 showed the exact point is stable under a plain retrain at L1 = 0. This cell adds the two
comparisons that complete the picture, all at H = 1 on det_init, tracking the off-manifold rule
deviation over training epochs. It reuses the section-5 training loop and adds the one thing that loop
left out: the L1 penalty on the recurrent weights, the term the committed fits actually used.

- **From the exact weights, L1 = 0.** The rule holds. Cross-entropy is content to sit at the exact
  solution, so training stays and the off-manifold deviation stays near the floor.
- **From the exact weights, L1 = 1e-5.** The same penalty the committed fits used. It pulls the
  network off the solution: holding win-stay needs large recurrent weights to saturate the gates, and
  L1 charges for exactly that, so the deviation climbs while the on-data loss keeps dropping.
- **From a cold random start, L1 = 0.** Training learns the policy on the data, its loss falls like
  the others, but the off-manifold deviation never comes down. Nothing past the data pulls a fresh
  network toward the exact rule.

```python
# --- 5.3 -- three starting points: does training hold the exact rule off the data? (cached) ---
# Reuses the section-5 harness (to_torch / from_torch / batch / rule_dev) at the same optimiser
# settings as the committed fits: AdamW at lr=0.005, and an L1 penalty on the input AND recurrent
# weights, the term tinyRNN uses. Three starts at H=1 on det_init: the exact solve with L1 off, the
# exact solve with L1=1e-5, and a cold start from torch's default init (best-learning of a few seeds).
def train_track_l1(d, P0, lr, epochs, every, n_train, eval_sess, l1):
    A, Y = d["A"], d["Y"]; X, Yt = batch(d, n_train); m = to_torch(P0, A, Y).double(); X = X.double()
    opt = torch.optim.AdamW(m.parameters(), lr=lr, weight_decay=0.0); lossf = nn.CrossEntropyLoss()
    ep, ce, rd = [], [], []
    for e in range(epochs + 1):
        if e % every == 0:
            Pn = from_torch(m); ep.append(e); ce.append(ce_rnn(d, Pn, eval_sess)); rd.append(rule_dev(d, Pn))
        opt.zero_grad()
        loss = lossf(m(X).reshape(-1, A), Yt.reshape(-1)) \
             + l1 * (m.rnn.weight_ih_l0.abs().sum() + m.rnn.weight_hh_l0.abs().sum())
        loss.backward(); opt.step()
    return np.array(ep), np.array(ce), np.array(rd)

def cold_start_P(seed, H, d):
    torch.manual_seed(seed)
    return from_torch(TinyGRU(H, d["A"], d["Y"]).double())

R3_H, R3_EPOCHS, R3_EVERY, R3_TRAIN, R3_EVAL, R3_LR = 1, 2000, 100, 200, 60, 0.005
R3_COLD_SEEDS = [0, 1, 2, 3, 4]
_d3 = DATA["det_init"]; _eval3 = _d3["SESS"][:R3_EVAL]; P3_exact = SOL["det_init"][R3_H][0]

R3_CACHE = os.path.join(REPO, "fitted_models", "_reachability3_cache.pkl")
R3_SIG = dict(H=R3_H, ep=R3_EPOCHS, every=R3_EVERY, n=R3_TRAIN, ev=R3_EVAL, lr=R3_LR, seeds=R3_COLD_SEEDS, v=2)
R3 = None
if os.path.exists(R3_CACHE):
    _b = pickle.load(open(R3_CACHE, "rb"))
    if _b.get("sig") == R3_SIG: R3 = _b["runs"]
if R3 is None:
    print("training three starts at H=1 on det_init (cached after this) ...")
    exact0 = train_track_l1(_d3, P3_exact, R3_LR, R3_EPOCHS, R3_EVERY, R3_TRAIN, _eval3, 0.0)
    exact1 = train_track_l1(_d3, P3_exact, R3_LR, R3_EPOCHS, R3_EVERY, R3_TRAIN, _eval3, 1e-5)
    colds = {sd: train_track_l1(_d3, cold_start_P(sd, R3_H, _d3), R3_LR, R3_EPOCHS, R3_EVERY, R3_TRAIN, _eval3, 0.0)
             for sd in R3_COLD_SEEDS}
    best = min(colds, key=lambda sd: colds[sd][1][-1])   # lowest final CE = the one that learned the policy
    R3 = dict(exact0=exact0, exact1=exact1, cold=colds[best], best=int(best),
              n_learned=int(sum(colds[sd][1][-1] < _d3["ce_fsc"] + 0.05 for sd in R3_COLD_SEEDS)))
    pickle.dump(dict(sig=R3_SIG, runs=R3), open(R3_CACHE, "wb"))

runs3 = {"cold start, L1=0": R3["cold"], "from exact, L1=0": R3["exact0"], "from exact, L1=1e-5": R3["exact1"]}
col3 = {"cold start, L1=0": "#d1495b", "from exact, L1=0": "#2a9e8f", "from exact, L1=1e-5": "#8a5fd6"}
print(f"of {len(R3_COLD_SEEDS)} cold seeds, {R3['n_learned']} learned the policy; the curve uses the best (seed {R3['best']}).\n")

fig, axes = plt.subplots(1, 2, figsize=(11.5, 4.3))
for nm, (ep, ce, rd) in runs3.items():
    axes[0].plot(ep, np.clip(rd, 1e-6, None), "o-", color=col3[nm], ms=3, lw=1.8, label=nm)
    axes[1].plot(ep, ce - _d3["ce_fsc"], "-", color=col3[nm], lw=1.9, label=nm)
axes[0].axhline(1e-2, color="0.4", ls="--", lw=1.0)
axes[0].text(R3_EPOCHS, 1.15e-2, " rule broken above", fontsize=7.5, color="0.4", ha="right", va="bottom")
axes[0].set_yscale("log"); axes[0].set_ylabel("off-manifold rule deviation (log)")
axes[0].set_title("Does the rule hold off the data, over training?", fontsize=10); axes[0].legend(fontsize=8)
axes[1].axhline(0.0, color="0.4", ls="--", lw=1.0)
axes[1].set_ylabel("cross-entropy minus the controller's"); axes[1].set_title("On-data fit, over training", fontsize=10)
axes[1].legend(fontsize=8)
for ax in axes: ax.set_xlabel("training epoch")
fig.suptitle("det_init, H=1: from the exact weights L1=0 holds the rule, L1=1e-5 erodes it, a cold start never finds it",
             fontsize=10)
plt.tight_layout(); plt.show()

print("final off-manifold rule deviation  (0 = still the rule; large = drifted off it):")
for nm, (ep, ce, rd) in runs3.items():
    print(f"  {nm:>19}: deviation {rd[-1]:.2e}   on-data CE gap {ce[-1] - _d3['ce_fsc']:+.2e}")
```

### 5.4 The same penalty on the trained networks: what it buys and what it costs

§5.3 showed the L1 penalty pulling a network off the exact solution. Here is the practical version on
the committed fits. The det_init GRUs ship at two recurrent-weight L1 values, the 1e-5 baseline and a
heavier 1e-3. Loaded and compared at matched hidden size, so only the penalty differs, they read out
below on the metrics that matter next to the two the penalty is meant to help.

At matched size the heavier penalty does not lower the predictive loss (it raises it), and it slightly
lowers the win-stay and lose-shift rates. It does lower the participation ratio, the count of effective
dimensions, and it deepens the off-manifold drift, the same direction §5.3 and Finding 2 found. The
one thing it clearly improves sits outside the fit: on det_init it sharpens the automatic
size-selection statistic from p = 0.029 to p = 0.004 (the selection story is Finding 1 §11), a
property of the selector, not of the model's behavior.

So the honest reading of the L1 penalty, gathered here: it buys a tidier-looking model, fewer
effective dimensions and a more decisive size pick, and pays for that with a slightly worse fit on the
data and worse behavior off it. It does not make the network a better model of the behavior.

```python
# --- 5.4 -- the same L1 penalty on the TRAINED networks: what it buys and what it costs -------
# Load the committed det_init GRUs at the 1e-5 baseline and the heavier 1e-3, reduce each to numpy,
# and compare at matched hidden size. The pandas pickle shim and FSC path are already set in §0; here
# we borrow tinyRNN's loader the same way Finding 1 and Finding 2 do (import switch + os.chdir).
import shutil, sys, pandas as _pd
TINYRNN_ROOT = os.path.abspath("tinyRNN")
_COLLIDE = ("utils", "FSC", "MAPSO", "parametrizations", "environments", "DiscreteObs")
def _use_tinyrnn_path():
    if FSC_SRC in sys.path: sys.path.remove(FSC_SRC)
    if TINYRNN_ROOT not in sys.path: sys.path.insert(0, TINYRNN_ROOT)
    for m in list(sys.modules):
        if m.split(".")[0] in _COLLIDE: del sys.modules[m]
def _use_fsc_path():
    if TINYRNN_ROOT in sys.path: sys.path.remove(TINYRNN_ROOT)
    if FSC_SRC not in sys.path: sys.path.insert(0, FSC_SRC)
    for m in list(sys.modules):
        if m.split(".")[0] in _COLLIDE: del sys.modules[m]

def _extract_gru(agent):
    net = agent.model
    p = {k: v.detach().cpu().numpy() for k, v in net.rnn.named_parameters()}
    return {"W_ih": p["weight_ih_l0"], "W_hh": p["weight_hh_l0"], "b_ih": p["bias_ih_l0"],
            "b_hh": p["bias_hh_l0"], "lin_W": net.lin.weight.detach().cpu().numpy(),
            "lin_b": net.lin.bias.detach().cpu().numpy(), "H": net.hidden_dim}

def _load_trained_gru(dataset, H, A, Y, suffix=""):
    _use_tinyrnn_path()
    exp = f"exp_{dataset}" + suffix
    live = f"files/saved_model/{exp}"; arch = os.path.join(REPO, "fitted_models", dataset, "rnn", exp)
    cwd = os.getcwd(); os.chdir("tinyRNN")
    try:
        if os.path.isdir(arch) and not os.path.isdir(live): shutil.copytree(arch, live)
        from agents import Agent
        from analyzing_experiments.analyzing_perf import select_final_rnn_perf
        _, summary, _ = select_final_rnn_perf(exp, return_dim_est=True, verbose=False)
        rows = summary[summary["hidden_dim"] == H].sort_values("trainval_loss")
        cfg = {"agent_type": "RNN", "rnn_type": "GRU", "input_dim": A + Y, "hidden_dim": H,
               "output_dim": A, "device": "cpu", "output_h0": True, "trainable_h0": False,
               "readout_FC": True, "one_hot": False, "seed": 0}
        ag = Agent("RNN", config=cfg); ag.load(str(rows.iloc[0]["model_path"]))
        return _extract_gru(ag)
    finally:
        os.chdir(cwd)

_dd = load_dataset("datasets/wsls_actions_observations_det_init.npz")
_sess = _dd["sessions"][:400]; _A, _Y, _wi, _li = _dd["A"], _dd["Y"], _dd["win_idx"], _dd["lose_idx"]

def _trained_metrics(P):
    nll = 0.0; na = 0; sw = sl = 0.0; nw = nl = 0; states = []
    for s in _sess:
        h = np.zeros(P["H"])
        for t in range(len(s["actions"])):
            p = rnn_action_probs(h, P); a = int(s["actions"][t])
            nll += -np.log(max(p[a], 1e-12)); na += 1
            if t >= 1:
                if s["observations"][t-1] == _wi: nw += 1; sw += p[int(s["actions"][t-1])]
                elif s["observations"][t-1] == _li: nl += 1; sl += 1 - p[int(s["actions"][t-1])]
            h, _ = gru_step(h, encode_input(a, int(s["observations"][t]), _A, _Y), P); states.append(h.copy())
    if P["H"] == 1:
        pr = 1.0
    else:
        Hs = np.array(states); Hs = Hs - Hs.mean(0)
        ev = np.clip(np.linalg.eigvalsh(np.cov(Hs.T)), 0, None); pr = float((ev.sum()**2) / (np.sum(ev**2) + 1e-30))
    worst = 1.0
    for a0 in range(_A):
        h = np.zeros(P["H"]); x = encode_input(a0, _wi, _A, _Y)
        for _ in range(80): h, _ = gru_step(h, x, P)
        worst = min(worst, float(rnn_action_probs(h, P)[a0]))
    return dict(loss=nll/na, stay_win=sw/nw, shift_lose=sl/nl,
                first=float(rnn_action_probs(np.zeros(P["H"]), P)[0]), pr=pr, drift80=worst)

_rows = []
for H in [2, 8]:
    b = _trained_metrics(_load_trained_gru("det_init", H, _A, _Y, ""))
    hv = _trained_metrics(_load_trained_gru("det_init", H, _A, _Y, "_l1-0.001"))
    _rows.append(dict(H=H, L1="1e-5 (baseline)", **{k: round(v, 4) for k, v in b.items()}))
    _rows.append(dict(H=H, L1="1e-3 (heavier)",  **{k: round(v, 4) for k, v in hv.items()}))
_use_fsc_path()
_tab = _pd.DataFrame(_rows)[["H", "L1", "loss", "stay_win", "shift_lose", "first", "pr", "drift80"]]
print("det_init trained GRUs, baseline L1=1e-5 vs heavier L1=1e-3, compared at matched hidden size:\n")
print(_tab.to_string(index=False))
print("\nloss = mean negative log-likelihood per trial (lower is better).  stay_win / shift_lose = WSLS rates.")
print("first = P(first action).  pr = participation ratio (effective dimensions).  drift80 = worst-arm P(stay) at k=80.")
print("\nReading it: the heavier penalty raises the loss and lowers the rates a little (worse fit), lowers pr and")
print("worsens drift80 (tidier and more broken off the data). The only gain, the sharper size-selection p-value,")
print("is a property of the selector, not the behaviour.")
```

## 6. What this establishes

**1. A GRU can hold the controller exactly, at every size tested including one hidden unit.** The solve
hits the constraint points to about 1e-8, and replaying it through the plain numpy GRU confirms it
reproduces the policy and holds the off-manifold probe flat (§3). So the drift in Finding 2 is not a
capacity limit of the architecture. It is about how the weights were arrived at.

**2. The correspondence between the two memories is an affine map, and it exists exactly when
H >= M - 1** (§4). For the two-state machine one hidden unit suffices. For the three-state machine one
unit reproduces the behavior but cannot hold the state, because three points collapse onto a line; two
units recover it. Matching the hidden size to the number of states is not the requirement; one less is.

**3. Solving is not the same as being reachable.** From the exact point, training with the L1 penalty
erodes the rule and training without it holds; a cold start never finds it, because the loss has no
gradient off the data (§5.3; the fuller basin sweep is in the experiments notebooks). To
actually get the exact controller you have to change the problem: start near the solution, add data
that exercises the off-manifold behavior, or construct the weights directly as here. On the committed trained networks the same penalty (§5.4) lowers the effective-dimension count and sharpens the size pick, and at matched size leaves the loss no better, the win-stay and lose-shift rates slightly worse, and the off-manifold drift worse, so the tidiness it buys is not a better model.

**4. Every separation in this notebook comes from a measurement taken off the data or from the
internal state directly.** On the trials the data contains, the solved network, the trained networks,
and the controller read the same. That is the thread through all three findings: behavior on ordinary
trials cannot tell these models apart, and the questions worth asking live off the data.

**Where to take it next.** A machine whose memory cannot be read off the last trial, with M >= 3
states, where H >= M - 1 becomes a real constraint and whether training finds the solved solution
becomes a question with something at stake.
